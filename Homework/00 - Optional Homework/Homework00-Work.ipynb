These are the claims we want to show from this dataset

```python
df = pd.read_csv(DATA_PATH)
df.head()
```

```python
# 2/3 of gun deaths are suicide
number_of_suicides = df[(df["Intent"] == "Suicide")&(df["Gender"] == "None selected")&(df["Race"] == "None selected")&(df["Age"] == "None selected")]["Deaths"].iloc[0]
# This means zake dataset filter for rows wherw Intent ==..., Gender....., then take th deaths column, and give the ifrst value

total_deaths = df[(df["Intent"] == "None selected")&(df["Gender"] == "None selected")&(df["Race"] == "None selected")&(df["Age"] == "None selected")]["Deaths"].iloc[0]

fraction_of_suicides = number_of_suicides / total_deaths *100

print(number_of_suicides)
print(total_deaths)
print("The total percentage of suicides is %d percent." %(fraction_of_suicides))
```

```python
#More than 85 percent of suicide victims are male.
male_suicides = df[(df["Intent"] == "Suicide")&(df["Gender"] == "Male")&(df["Race"] == "None selected")&(df["Age"] == "None selected")]["Deaths"].iloc[0]

fraction_of_male_sc = male_suicides / number_of_suicides * 100
print("The total percentage of male suicides is %d percent." %(fraction_of_male_sc))
```

```python
# Around a third of all gun deaths are homicides.
number_of_h = df[(df["Intent"] == "Homicide")&(df["Gender"] == "None selected")&(df["Race"] == "None selected")&(df["Age"] == "None selected")]["Deaths"].iloc[0]
# This means zake dataset filter for rows wherw Intent ==..., Gender....., then take th deaths column, and give the ifrst value

total_deaths = df[(df["Intent"] == "None selected")&(df["Gender"] == "None selected")&(df["Race"] == "None selected")&(df["Age"] == "None selected")]["Deaths"].iloc[0]

fraction_of_h = number_of_h / total_deaths *100

print(number_of_suicides)
print(total_deaths)
print("The total percentage of homice is %d percent." %(fraction_of_h))
```
